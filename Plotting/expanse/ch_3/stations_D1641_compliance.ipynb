# update compliance classification for salinity at stations using D1641 rules

#From D-1641 footnote [2] of table 1: "Determination of compliance with an objective expressed as a running average begins on the last day of the averaging period. The averaging period commences with the first day of the time period for the applicable objective. If the objective is not met on the last day of the averaging period, all days in the averaging  period are considered out of compliance."

#Rules apply on a water year classification basis in table 1. In our study period, 2019 is W, 2020 is D, and 2021 is C

#stations in table 1:
CCWD maximum mean daily Cl- concentration must be below 150 mg/L for at least 
240 days in 2019
165 days in 2020 and 
155 days in 2021.
Maximum mean daily Cl- concentration cannot exclude 250 mg/L in any water year for any part of the year.

Emmaton: Maximum 14 day running average of mean daily EC needs to be below 0.45 mmhos/cm between April 1 and 
August 15 in 2019,
June 15 in 2020
Then in 2020 it must be below 1.67 mmhos/cm until August 15
and in 2021 in must be below 2.78 mmhos/cm April 1  August 15.

Jersey Point: Maximum 14 day running average of mean daily EC needs to be below 0.45 mmhos/cm between April 1 and 
August 15 in 2019 and 
June 15 in 2020.
Then in 2020 it must be below 1.35 mmhos/cm until August 15
and in 2021 in must be below 2.20 mmhos/cm from April 1 until August 15.

CCFB: Maximum monthly average of mean daily EC cannot exceed 1 mmhos/cm for the whole water year.
Maximum mean daily Cl- concentration cannot exclude 250 mg/L in any water year for any part of the year.
