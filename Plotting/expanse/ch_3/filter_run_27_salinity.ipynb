```python
import argparse
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
from vtools.functions.filter import cosine_lanczos
from vtools.data.vtime import days, hours, minutes
```

```python
def filter_timeseries(tss, cutoff_period=hours(40)):
    #function comes from schimpy metricsplot.py script https://github.com/CADWRDeltaModeling/schimpy/blob/master/schimpy/metricsplot.py
    """ Filter time series

        Parameters
        ----------

        Returns
        -------
        list of vtools.data.timeseries.TimeSeries
            filtered time series
    """

    filtered = []
    ts=tss #try removing the loop for now
    if ts is None:
        filtered.append(None)
    else:
        ts_filtered = cosine_lanczos(ts.astype("float64"), cutoff_period=cutoff_period)
        ts_filtered.filtered = 'cosine_lanczos'
    return ts_filtered
```

```python
filepath="/expanse/lustre/scratch/jisrael/temp_project/run_schism/run_27/station_files/run_27_salinity_padded.txt"
nstations=408
timebasis='08-14-2018T00:00:00' 
startfilter='09-01-2018T00:00:00'
```

```python
columns = ['time']
#stations_no = [str(i) for i in range(nstations)]
#EDIT 11/18/2025: station columns should start with 1 NOT 0 to be consistent with the raw station data/ station.in file numbering 
stations_no = [str(i) for i in np.arange(1,nstations+1)]
columns = columns + stations_no
station = pd.read_csv(filepath, sep='\\s+', names=columns,dtype=np.float32)
station
```

```python
#fill nans
station = station.fillna(-9999)
timedelta = [pd.Timedelta(i, unit = 's') for i in station.time]
timestamp = pd.Series([pd.to_datetime(timebasis) + i for i in timedelta])
timestamp
```

```python
timestamp.dt
```

```python
#added 07/01 on Savio
roundtime=timestamp.dt.round(freq='900s')
station['time'] = roundtime
station.set_index("time",inplace=True)
station=station.asfreq('900s')
```

```python
#update 11/18/2025: trim the data to where you want to start filtering(e.g. after the nudging/ model warm-up), due to the 40 hr cosine filter end up losing the first 100 hours (400 samples of 15 minute data) so actually trim 40 hours back of this date
fstart=pd.to_datetime(startfilter)-pd.Timedelta(hours=100)
if fstart < pd.to_datetime(timebasis):
    print('Need to choose later time to start filtering')
station2filt=station[fstart:]
```
